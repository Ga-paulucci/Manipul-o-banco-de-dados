Importando bibliotecas e fazendo leitura do banco.

```python
import pandas as pd
import sqlalchemy
dados_desafio = 'https://raw.githubusercontent.com/alura-cursos/Pandas/main/clientes_banco.csv'
pd.read_csv(dados_desafio)
```

Importando funções, criando banco local e dando nome para a tabela.

```python
from sqlalchemy import MetaData, create_engine, inspect, Table
```

```python
engine = create_engine('sqlite:///:memory:')
inspec = inspect(engine)
df_clientes_desafio = pd.read_csv(dados_desafio)
df_clientes_desafio.to_sql('clientes_desafio', engine, index=False)
```

```python
print(inspec.get_table_names())
```

Importando a biblioteca text para leitura direta das querys sql.

```python
from sqlalchemy import text
```

Query para atualização do rendimento anual do cliente.

```python
query = 'UPDATE clientes_desafio SET Rendimento_anual = 300000 WHERE ID_Cliente = 6840104'
with engine.connect() as conn:
  result = conn.execute(text(query))
  conn.commit()
```

Consulta para confirmação da atualização do cliente.

```python
query = 'SELECT * FROM clientes_desafio WHERE ID_Cliente = 6840104'
pd.read_sql(query, engine)
```

Query para deletar cliente do banco de dados.

```python
query = 'DELETE FROM clientes_desafio WHERE ID_Cliente = 5008809'
with engine.connect() as conn:
  result = conn.execute(text(query))
  conn.commit()
```

Consulta para confirmar o delete do cliente.

```python
query = 'SELECT * FROM clientes_desafio WHERE ID_Cliente = 5008809'
pd.read_sql(query, engine)
```

Query para adicionar um cliente na tabela.

```python
query = 'INSERT INTO clientes_desafio (ID_Cliente, Idade, Grau_escolaridade, Estado_civil, Tamanho_familia, Categoria_de_renda, Ocupacao, Anos_empregado, Rendimento_anual, Tem_carro, Moradia) VALUES (6850985, 33, "Doutorado", "Solteiro", 1, "Empregado", "TI", 2, 290000, 0, "Casa/apartamento próprio")'
with engine.connect() as conn:
  result = conn.execute(text(query))
  conn.commit()
```
